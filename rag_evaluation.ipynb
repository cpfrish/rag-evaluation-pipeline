### Doc loader functions

```python
# Document loaders for Arxiv, Wikipedia, and blogs.
#
# Each loader:
#   - prints per-item progress so a stall is visible in real time
#   - uses explicit HTTP timeouts so one dead URL can't hang the notebook
#   - isolates per-item failures (a bad PDF skips instead of killing the run)
#   - caches raw arxiv PDFs on disk so reruns skip the network entirely

# decided to organize them here instead of in a separate file to keep all the data loading logic in one place for easier debugging and modification during development.
import shutil
from pathlib import Path as _Path

import requests

ARXIV_PDF_CACHE = _Path("arxiv_pdfs")
HTTP_TIMEOUT = 30  # seconds per request


def _download_arxiv_pdf(
    arxiv_id: str, dest_dir: _Path, timeout: int = HTTP_TIMEOUT
) -> _Path | None:
    """Download an arxiv PDF to dest_dir; return the local path or None on failure."""
    dest_dir.mkdir(parents=True, exist_ok=True)
    dest = dest_dir / f"{arxiv_id}.pdf"
    if dest.exists() and dest.stat().st_size > 0:
        return dest

    url = f"https://arxiv.org/pdf/{arxiv_id}.pdf"
    try:
        with requests.get(url, stream=True, timeout=timeout) as r:
            r.raise_for_status()
            tmp = dest.with_suffix(".pdf.part")
            with open(tmp, "wb") as f:
                shutil.copyfileobj(r.raw, f)
            tmp.rename(dest)
        return dest
    except Exception as e:
        print(f"! arxiv {arxiv_id} download failed: {e}")
        if dest.exists():
            dest.unlink()
        return None


def load_arxiv_documents(
    arxiv_ids: list,
    text_splitter: RecursiveCharacterTextSplitter,
    load_max_docs: int = 1,
    verbose: bool = True,
    cache_dir: _Path = ARXIV_PDF_CACHE,
) -> list:
    docs = []
    total = len(arxiv_ids)
    for i, arxiv_id in enumerate(arxiv_ids, start=1):
        if verbose:
            print(f"  [{i}/{total}] arxiv:{arxiv_id}", flush=True)
        pdf_path = _download_arxiv_pdf(arxiv_id, cache_dir)
        if pdf_path is None:
            continue
        try:
            loaded = PyMuPDFLoader(str(pdf_path)).load()
        except Exception as e:
            print(f"    ! arxiv {arxiv_id} parse failed: {e}")
            continue
        docs.extend(loaded[:load_max_docs] if load_max_docs else loaded)

    if not docs:
        if verbose:
            print("Arxiv: 0 docs loaded.")
        return []

    splits = text_splitter.split_documents(docs)
    for idx, chunk in enumerate(splits):
        chunk.metadata["split_id"] = idx

    if verbose:
        print(f"Arxiv docs={len(docs)} splits={len(splits)}")
    return splits


def load_wiki_documents(
    wiki_queries: dict,
    text_splitter: RecursiveCharacterTextSplitter,
    verbose: bool = True,
) -> list:
    docs = []
    total = len(wiki_queries)
    for i, (query, max_docs) in enumerate(wiki_queries.items(), start=1):
        if verbose:
            print(f"  [{i}/{total}] wiki:{query!r}", flush=True)
        try:
            loaded = WikipediaLoader(query=query, load_max_docs=max_docs).load()
        except Exception as e:
            print(f"    ! wiki {query!r} failed: {e}")
            continue
        docs.extend(loaded)

    if not docs:
        if verbose:
            print("Wiki: 0 docs loaded.")
        return []

    splits = text_splitter.split_documents(docs)
    for idx, split in enumerate(splits):
        split.metadata["split_id"] = idx

    if verbose:
        print(f"Wiki docs={len(docs)} splits={len(splits)}")
    return splits


def load_blog_documents(
    document_paths: list,
    text_splitter: RecursiveCharacterTextSplitter,
    user_agent: str | None = None,
    verbose: bool = True,
) -> list:
    docs = []
    total = len(document_paths)
    for i, path in enumerate(document_paths, start=1):
        if verbose:
            print(f"  [{i}/{total}] blog:{path}", flush=True)
        kwargs = {
            "web_path": path,
            "bs_kwargs": {
                "parse_only": bs4.SoupStrainer(
                    class_=("post-content", "post-title", "post-header")
                )
            },
            "requests_kwargs": {"timeout": HTTP_TIMEOUT},
        }
        if user_agent:
            kwargs["header_template"] = {"User-Agent": user_agent}
        try:
            loader = WebBaseLoader(**kwargs)
            loaded = loader.load()
        except Exception as e:
            print(f"    ! blog {path} failed: {e}")
            continue
        docs.extend(loaded)

    if not docs:
        if verbose:
            print("Blogs: 0 docs loaded.")
        return []

    for idx, doc in enumerate(docs):
        doc.metadata["doc_num"] = idx
        doc.metadata["doc_source"] = "WWW"

    splits = text_splitter.split_documents(docs)
    for idx, split in enumerate(splits):
        split.metadata["split_id"] = idx

    if verbose:
        print(f"Blogs docs={len(docs)} splits={len(splits)}")
    return splits
```

## 2s Embedding Model


```python
# given models we can use
embedding_models = [
    "all-MiniLM-L6-v2",
    "multi-qa-mpnet-base-dot-v1",
    "all-mpnet-base-v2",
    "all-distilroberta-v1",
    "multi-qa-distilbert-cos-v1",
]

# one of many functions I had to write to work around HF downloading issues in Colab + running locally.


def _ensure_hf_model(repo_id: str, max_retries: int = 4) -> None:
    """
    Pre-download a HuggingFace model with resume-on-failure.

    This runs before HuggingFaceEmbeddings() so the library loads from cache
    instead of doing its own (non-resumable) download path.
    """
    from huggingface_hub import snapshot_download
    from huggingface_hub.utils import HfHubHTTPError

    candidates = [repo_id]
    if "/" not in repo_id:
        candidates.append(f"sentence-transformers/{repo_id}")

    last_err = None
    for candidate in candidates:
        for attempt in range(1, max_retries + 1):
            try:
                snapshot_download(
                    repo_id=candidate,
                    max_workers=2,
                    # Skip duplicate format variants (TF, Flax, ONNX, .bin when
                    # safetensors exists). avoids
                    # the runtime-disconnect on long downloads which happened to me several times.
                    allow_patterns=[
                        "*.safetensors",
                        "*.json",
                        "*.txt",
                        "tokenizer*",
                        "vocab*",
                        "merges*",
                        "spiece*",
                        "sentence_bert_config*",
                        "modules.json",
                        "1_Pooling/*",
                        "2_Normalize/*",
                    ],
                )
                return
            except HfHubHTTPError as e:
                # 401/404 means wrong repo — try the next candidate
                last_err = e
                if getattr(e.response, "status_code", None) in (401, 404):
                    break
                print(f"  ! {candidate} attempt {attempt}/{max_retries} failed: {e}")
            except Exception as e:
                last_err = e
                print(f"  ! {candidate} attempt {attempt}/{max_retries} failed: {e}")
    raise RuntimeError(
        f"Could not download HF model {repo_id!r} after retries: {last_err}"
    )


def generate_embeddings(model_name: str) -> HuggingFaceEmbeddings:
    """Generate embeddings for a given sentence-transformers model name."""
    _ensure_hf_model(model_name)
    return HuggingFaceEmbeddings(model_name=model_name)


# function to compute cosine similarity between a query embedding and document embeddings
# will use this to rank our documents based on their relevance to the query
def compute_similarity(query_result, doc_result):
    """Compute cosine similarity between query and document embeddings."""
    query_array = np.array(query_result).reshape(1, -1)
    doc_array = np.array(doc_result)
    similarity = cosine_similarity(query_array, doc_array)[0]
    return similarity
```

## 3s Vector Store

```python
def load_and_prepare_documents(
    arxiv_ids: list[str],
    wiki_queries: dict,
    blogs: list,
    text_splitter: RecursiveCharacterTextSplitter,
    user_agent: str | None = None,
    verbose: bool = False,
) -> list:
    """Load and split documents from Arxiv, Wikipedia, and blog sources."""
    arxiv_docs = load_arxiv_documents(arxiv_ids, text_splitter, verbose=verbose)
    wiki_docs_list = load_wiki_documents(wiki_queries, text_splitter, verbose=verbose)
    blog_docs = load_blog_documents(
        blogs,
        text_splitter,
        user_agent=user_agent,
        verbose=verbose,
    )
    return arxiv_docs + wiki_docs_list + blog_docs


def prepare_vectorstore(
    documents: list,
    embedding_model: HuggingFaceEmbeddings,
) -> QdrantVectorStore:
    """Build in-memory Qdrant vector store from document chunks and embeddings."""
    return QdrantVectorStore.from_documents(
        documents=documents,
        embedding=embedding_model,
        location=":memory:",
        collection_name="rag_tech_db",
        force_recreate=True,
    )


def load_and_prepare_vectorstore(
    arxiv_ids: list[str],
    wiki_queries: dict,
    blogs: list,
    text_splitter: RecursiveCharacterTextSplitter,
    embedding_model: HuggingFaceEmbeddings,
    user_agent: str | None = None,
    verbose: bool = False,
) -> QdrantVectorStore:
    """Load docs, split, embed, then return a vector store ready for retrieval."""
    documents = load_and_prepare_documents(
        arxiv_ids,
        wiki_queries,
        blogs,
        text_splitter,
        user_agent=user_agent,
        verbose=verbose,
    )
    return prepare_vectorstore(documents, embedding_model)
```

```python
# Build/load document chunks with a Colab-friendly cache flow. Originally I was using this locally and it was working well, but when I moved to Colab I ran into issues with the default Hugging Face downloader hanging on large files due to network instability. This cache flow allowed me to pre-download the necessary files and load them from disk, avoiding the network issues during the main execution of the notebook. It also provides a way to upload cached files directly in Colab if they are not found.
import pickle
from pathlib import Path as _P

# Cache helper locations for local + Colab runtimes
DOC_CACHE_SEARCH_DIRS = [
    _P("."),
    _P("/content"),
    _P("/content/drive/MyDrive"),
    _P("/content/drive/MyDrive/Assignment_5"),
    _P.home(),
]


def resolve_doc_cache(filename: str, allow_upload: bool = True):
    """Return a Path to an existing cache file, or None if not found."""
    name = _P(filename).name
    for d in DOC_CACHE_SEARCH_DIRS:
        candidate = d / name
        if candidate.exists() and candidate.stat().st_size > 0:
            return candidate
    if allow_upload and IS_COLAB:
        try:
            from google.colab import files  # type: ignore

            print(f"Cache {name!r} not found. Upload it now (cancel to rebuild):")
            uploaded = files.upload()
            for uploaded_name in uploaded:
                if uploaded_name.endswith(name) or uploaded_name == name:
                    dest = _P("/content") / uploaded_name
                    if dest.exists():
                        return dest
        except Exception as e:
            print(f"Colab upload skipped: {e}")
    return None


def save_doc_cache(filename: str, obj) -> _P:
    """Write cache to cwd and return the path."""
    dest = _P(filename).name
    with open(dest, "wb") as f:
        pickle.dump(obj, f)
    return _P(dest)


text_splitter = create_text_splitter(chunk_size=1000, chunk_overlap=200)
embeddings = generate_embeddings("all-mpnet-base-v2")

CACHE_NAME = "cached_documents.pkl"
cache_path = resolve_doc_cache(CACHE_NAME)

if cache_path is not None:
    print(f"Loading documents from cache: {cache_path}")
    with open(cache_path, "rb") as f:
        all_docs = pickle.load(f)
    print(f"Loaded {len(all_docs)} cached chunks.")
else:
    print("No cache found. Downloading and chunking documents (first run)...")
    print("\n-- Arxiv --", flush=True)
    arxiv_chunks = load_arxiv_documents(arxiv_ids, text_splitter)
    print("\n-- Wikipedia --", flush=True)
    wiki_chunks = load_wiki_documents(wiki_docs, text_splitter)
    print("\n-- Blogs --", flush=True)
    blog_chunks = load_blog_documents(blogs, text_splitter, user_agent=USER_AGENT)
    all_docs = arxiv_chunks + wiki_chunks + blog_chunks
    saved = save_doc_cache(CACHE_NAME, all_docs)
    print(f"\nCached {len(all_docs)} chunks to {saved}")

# Note: build_rag_pipeline() creates per-config embeddings/vectorstores separately.
qdrant_vectorstore = prepare_vectorstore(all_docs, embeddings)
```

```python
def evaluate_rag(dataset: dict, graph, scorer=None):
    """
    Evaluate a built RAG pipeline with BERTScore, ROUGE, faithfulness, and length ratio.

    Faithfulness is computed against `extracted_facts` (what the generators
    actually received), not the raw retrieved context. This makes the metric
    a better grounding check on the generator's input.

    Returns
    -------
    all_metrics : dict  {persona: {bertscore, rouge, length_ratio, faithfulness_f1}}
    raw_rows    : list[dict]  one entry per question with gold, generated, and
                              per-question bert_f1_research / bert_f1_marketing
    """
    preds_res, refs_res = [], []
    preds_mkt, refs_mkt = [], []
    facts, item_ids, questions = [], [], []

    def _scalar(t):
        if t is None:
            return None
        if hasattr(t, "mean"):
            t = t.mean()
        if hasattr(t, "item"):
            return float(t.item())
        return float(t)

    # 1. Single inference pass
    for item_id, item in dataset.items():
        query = item.get("question", "")
        refs_res.append(item.get("gold_answer_research") or "")
        refs_mkt.append(item.get("gold_answer_marketing") or "")
        item_ids.append(item_id)
        questions.append(query)

        result = graph.invoke({"question": query})
        preds_res.append(result.get("research_answer") or "")
        preds_mkt.append(result.get("marketing_answer") or "")
        facts.append(
            result.get("extracted_facts") or result.get("retrieved_context") or ""
        )

    def _rouge_avg(preds, refs):
        r1, r2, rl = [], [], []
        for p, r in zip(preds, refs):
            s = _rouge_scorer.score(r, p)
            r1.append(s["rouge1"].fmeasure)
            r2.append(s["rouge2"].fmeasure)
            rl.append(s["rougeL"].fmeasure)
        return float(np.mean(r1)), float(np.mean(r2)), float(np.mean(rl))

    def _length_ratio(preds, refs):
        ratios = [len(p.split()) / max(len(r.split()), 1) for p, r in zip(preds, refs)]
        return float(np.mean(ratios))

    def _bert(preds, refs):
        if scorer is not None:
            return scorer.score(preds, refs)
        from bert_score import score as bscore

        return bscore(preds, refs, lang="en", verbose=False)

    # 2. BERTScore vs gold responses (per persona) and vs extracted facts (faithfulness)
    p_res, r_res, f1_res = _bert(preds_res, refs_res)
    p_mkt, r_mkt, f1_mkt = _bert(preds_mkt, refs_mkt)

    if any(f for f in facts):
        _, _, ff1_res = _bert(preds_res, facts)
        _, _, ff1_mkt = _bert(preds_mkt, facts)
    else:
        ff1_res = ff1_mkt = None

    # 3. Per-question rows
    raw_rows = []
    for i, (item_id, question) in enumerate(zip(item_ids, questions)):
        raw_rows.append(
            {
                "id": item_id,
                "question": question,
                "gold_research": refs_res[i],
                "gold_marketing": refs_mkt[i],
                "gen_research": preds_res[i],
                "gen_marketing": preds_mkt[i],
                "extracted_facts": facts[i],
                "bert_f1_research": _scalar(f1_res[i]),
                "bert_f1_marketing": _scalar(f1_mkt[i]),
                "rouge_l_research": _rouge_scorer.score(refs_res[i], preds_res[i])[
                    "rougeL"
                ].fmeasure,
                "rouge_l_marketing": _rouge_scorer.score(refs_mkt[i], preds_mkt[i])[
                    "rougeL"
                ].fmeasure,
            }
        )

    # 4. Aggregate metrics
    all_metrics = {}
    for persona, preds, refs, p_t, r_t, f1_t, ff1_t in [
        ("Research", preds_res, refs_res, p_res, r_res, f1_res, ff1_res),
        ("Marketing", preds_mkt, refs_mkt, p_mkt, r_mkt, f1_mkt, ff1_mkt),
    ]:
        r1, r2, rl = _rouge_avg(preds, refs)
        all_metrics[persona] = {
            "bertscore": (_scalar(p_t), _scalar(r_t), _scalar(f1_t)),
            "rouge": (r1, r2, rl),
            "length_ratio": _length_ratio(preds, refs),
            "faithfulness_f1": _scalar(ff1_t),
        }

    return all_metrics, raw_rows
```

```python
import hashlib
from dataclasses import dataclass, field, replace


DEFAULT_PROMPTS = {
    "rewriter": REWRITER_PROMPT,
    "reader": READER_PROMPT,
    "system": SYSTEM_PROMPT_TEMPLATE,
    "research": RESEARCH_PROMPT_TEMPLATE,
    "marketing": MARKETING_PROMPT_TEMPLATE,
}


@dataclass
class RAGConfig:
    """Core tunable settings for the RAG pipeline."""

    # Identity
    name: str = "default"

    # Embedding + chunking
    embedding_model: str = "all-mpnet-base-v2"
    chunk_size: int = 1000
    chunk_overlap: int = 200

    # LLM + quantization
    llm_provider: str = "huggingface"  # "huggingface" or "cohere"
    llm_model: str = "mistralai/Mistral-7B-Instruct-v0.3"
    cohere_model: str = "command-a-03-2025"
    load_in_4bit: bool = True
    load_in_8bit: bool = False
    bnb_4bit_quant_type: str = "nf4"
    bnb_4bit_use_double_quant: bool = True
    bnb_4bit_compute_dtype: str = "bfloat16"

    # Generation
    max_new_tokens: int = 384
    temperature: float = 0.1
    top_p: float = 0.95
    do_sample: bool = True
    repetition_penalty: float | None = None

    # Retrieval
    retriever_k: int = 5
    reranker_model: str = "BAAI/bge-reranker-v2-m3"
    reranker_top_n: int = 3
    use_rewriter: bool = True
    use_reranker: bool = True

    # Prompt tweaks
    prompt_overrides: dict[str, str] = field(default_factory=dict)

    # Data
    arxiv_ids: list = field(default_factory=lambda: arxiv_ids)
    wiki_queries: dict = field(default_factory=lambda: wiki_docs)
    blogs: list = field(default_factory=lambda: blogs)

    # Cache
    save_doc_cache: bool = True
    cache_prefix: str = "cached_docs"

    @property
    def cache_key(self) -> str:
        key_parts = (
            tuple(self.arxiv_ids),
            tuple(sorted(self.wiki_queries.items())),
            tuple(self.blogs),
            self.chunk_size,
            self.chunk_overlap,
        )
        key_hash = hashlib.md5(str(key_parts).encode()).hexdigest()[:10]
        return f"{self.cache_prefix}_{self.name}_{key_hash}.pkl"

    @property
    def compute_dtype(self):
        return (
            torch.bfloat16
            if self.bnb_4bit_compute_dtype == "bfloat16"
            else torch.float16
        )

    def validate(self):
        if self.llm_provider not in {"huggingface", "cohere"}:
            raise ValueError("llm_provider must be 'huggingface' or 'cohere'.")
        if self.load_in_4bit and self.load_in_8bit:
            raise ValueError("Only one quantization mode can be active at a time.")
        if self.reranker_top_n > self.retriever_k:
            raise ValueError("reranker_top_n cannot exceed retriever_k.")

    def resolved_prompts(self) -> dict[str, str]:
        prompts = dict(DEFAULT_PROMPTS)
        prompts.update(self.prompt_overrides)
        return prompts

    def with_updates(self, **kwargs):
        cfg = replace(self, **kwargs)
        if "prompt_overrides" in kwargs and kwargs["prompt_overrides"] is not None:
            cfg.prompt_overrides = dict(kwargs["prompt_overrides"])
        return cfg

    def summary(self) -> str:
        if self.llm_provider == "cohere":
            quant = "api"
            llm_label = self.cohere_model
        else:
            quant = (
                "4bit"
                if self.load_in_4bit
                else ("8bit" if self.load_in_8bit else "none")
            )
            llm_label = self.llm_model.split("/")[-1]
        return (
            f"{self.name}: embed={self.embedding_model}, "
            f"chunk={self.chunk_size}/{self.chunk_overlap}, "
            f"llm={self.llm_provider}:{llm_label}, "
            f"quant={quant}, temp={self.temperature}, "
            f"max_tok={self.max_new_tokens}, rewriter={self.use_rewriter}, "
            f"reranker={self.use_reranker}, k={self.retriever_k}"
        )

    def __repr__(self):
        return self.summary()


def make_config(name: str, base: RAGConfig | None = None, **overrides) -> RAGConfig:
    base_cfg = base if base is not None else RAGConfig()
    cfg = base_cfg.with_updates(name=name, **overrides)
    cfg.validate()
    return cfg
```

```python
# ---- Cohere API Rate Limiting Wrapper ----
# Given my runs were taking SO long, getting a failure after 20 requests and losing progress was a real pain. This wrapper was my solution to that problem, allowing me to run the notebook end-to-end with the Cohere API without hitting silent stalls or losing progress due to rate limit errors. It implements a process-wide sliding window limiter to stay under the 20 req/min cap, and also includes an exponential backoff retry strategy if a 429 error does slip through.

# Rate-limited Cohere wrapper for the trial-key 20 req/min cap.
# Wraps ChatCohere with:
#   1. A process-wide sliding-window limiter (default 18 req/60s, leaves headroom)
#   2. Retry-with-backoff on TooManyRequestsError (429)
import threading
import time as _time
from collections import deque

try:
    from cohere.errors import TooManyRequestsError as _CohereRateLimit
except Exception:  # pragma: no cover - fallback if module path changes
    _CohereRateLimit = Exception


class _CohereRateLimiter:
    """Sliding-window limiter shared across every RateLimitedChatCohere instance."""

    def __init__(self, max_calls: int = 18, window_seconds: float = 60.0):
        self.max_calls = max_calls
        self.window = window_seconds
        self._calls: deque[float] = deque()
        self._lock = threading.Lock()

    def acquire(self) -> None:
        while True:
            with self._lock:
                now = _time.monotonic()
                while self._calls and now - self._calls[0] >= self.window:
                    self._calls.popleft()
                if len(self._calls) < self.max_calls:
                    self._calls.append(now)
                    return
                wait = self.window - (now - self._calls[0]) + 0.05
            print(f"  [cohere-limiter] sleeping {wait:.1f}s to respect rate limit")
            _time.sleep(max(wait, 0.1))


COHERE_LIMITER = _CohereRateLimiter(max_calls=18, window_seconds=60.0)


class RateLimitedChatCohere(ChatCohere):
    """ChatCohere subclass that throttles + retries the generate path."""

    def _generate(self, *args, **kwargs):
        max_attempts = 6
        for attempt in range(max_attempts):
            COHERE_LIMITER.acquire()
            try:
                return super()._generate(*args, **kwargs)
            except _CohereRateLimit as e:
                if attempt == max_attempts - 1:
                    raise
                backoff = min(60.0, 2**attempt + 1.0)
                print(
                    f"  [cohere-limiter] 429 on attempt {attempt + 1}; "
                    f"sleeping {backoff:.1f}s before retry"
                )
                _time.sleep(backoff)
        raise RuntimeError("RateLimitedChatCohere: exhausted retries")
```

This is where all the pieces come together: document loading (with cache), vectorstore prep, LLM loading (with quantization), optional reranker setup, and then graph compilation. The function returns both the compiled graph and a dict of the internal components for easy inspection and debugging.

```python
def build_rag_pipeline(config: RAGConfig):
    """
    Factory that builds a complete RAG pipeline from a RAGConfig.
    Returns (compiled_graph, components_dict) to easily inspect internals.
    """
    import pickle

    config.validate()
    prompts = config.resolved_prompts()

    print(f"Building pipeline: {config.summary()}")
    print("-" * 60)

    # 1. Embeddings & chunking
    print("  [1/6] Creating embeddings & text splitter...")
    cfg_embeddings = generate_embeddings(config.embedding_model)
    cfg_splitter = create_text_splitter(config.chunk_size, config.chunk_overlap)

    # 2. Documents (cached)
    print("  [2/6] Loading documents...")
    cache_name = config.cache_key
    cfg_docs = None
    cache_resolver = globals().get("resolve_doc_cache")
    cache_saver = globals().get("save_doc_cache")

    if config.save_doc_cache and callable(cache_resolver):
        found = cache_resolver(cache_name, allow_upload=False)
        if found is not None:
            with open(str(found), "rb") as f:
                cfg_docs = pickle.load(f)
            print(f"         Loaded {len(cfg_docs)} chunks from {found}.")

    if cfg_docs is None:
        cfg_docs = load_and_prepare_documents(
            config.arxiv_ids,
            config.wiki_queries,
            config.blogs,
            cfg_splitter,
            user_agent=USER_AGENT,
            verbose=True,
        )
        if config.save_doc_cache and callable(cache_saver):
            saved = cache_saver(cache_name, cfg_docs)
            print(f"         Cached {len(cfg_docs)} chunks to {saved}")

    # 3. Vectorstore
    print("  [3/6] Building vectorstore...")
    cfg_vectorstore = prepare_vectorstore(cfg_docs, cfg_embeddings)

    # 4. LLM
    print("  [4/6] Loading LLM...")
    cfg_model = None

    if config.llm_provider == "huggingface":
        if torch.cuda.is_available():
            if config.load_in_4bit:
                qconfig = BitsAndBytesConfig(
                    load_in_4bit=True,
                    bnb_4bit_quant_type=config.bnb_4bit_quant_type,
                    bnb_4bit_use_double_quant=config.bnb_4bit_use_double_quant,
                    bnb_4bit_compute_dtype=config.compute_dtype,
                )
            elif config.load_in_8bit:
                qconfig = BitsAndBytesConfig(load_in_8bit=True)
            else:
                qconfig = None
        else:
            qconfig = None

        _ensure_hf_model(config.llm_model)
        cfg_model = AutoModelForCausalLM.from_pretrained(
            config.llm_model,
            device_map="auto",
            dtype=config.compute_dtype,
            quantization_config=qconfig,
        )
        cfg_tokenizer = AutoTokenizer.from_pretrained(config.llm_model)

        if cfg_tokenizer.pad_token_id is None:
            cfg_tokenizer.pad_token = cfg_tokenizer.eos_token
        cfg_model.config.pad_token_id = cfg_tokenizer.pad_token_id

        cfg_generation_config = GenerationConfig.from_model_config(cfg_model.config)
        cfg_generation_config.max_new_tokens = config.max_new_tokens
        cfg_generation_config.temperature = config.temperature
        cfg_generation_config.top_p = config.top_p
        cfg_generation_config.do_sample = config.do_sample
        cfg_generation_config.pad_token_id = cfg_tokenizer.pad_token_id
        cfg_generation_config.eos_token_id = cfg_tokenizer.eos_token_id
        if config.repetition_penalty is not None:
            cfg_generation_config.repetition_penalty = config.repetition_penalty

        cfg_pipe = pipeline(
            task="text-generation",
            model=cfg_model,
            tokenizer=cfg_tokenizer,
            dtype="auto",
        )
        cfg_llm = HuggingFacePipeline(
            pipeline=cfg_pipe,
            pipeline_kwargs={
                "generation_config": cfg_generation_config,
                "return_full_text": False,
            },
        )
        mem_gb = round(cfg_model.get_memory_footprint() / (1024**3), 4)
        print(f"         Model memory: {mem_gb} GB")

    else:
        if not COHERE_API_KEY:
            raise ValueError(
                "COHERE_API_KEY is not set. Add it to your environment before running Cohere configs."
            )
        cfg_llm = RateLimitedChatCohere(
            model=config.cohere_model,
            cohere_api_key=COHERE_API_KEY,
            temperature=config.temperature,
            max_tokens=config.max_new_tokens,
        )
        print(f"         Cohere model: {config.cohere_model}")

    # 5. Reranker
    cfg_compressor = None
    if config.use_reranker:
        print("  [5/6] Setting up reranker...")
        _ensure_hf_model(config.reranker_model)
        cfg_reranker_model = HuggingFaceCrossEncoder(model_name=config.reranker_model)
        cfg_compressor = CrossEncoderReranker(
            model=cfg_reranker_model, top_n=config.reranker_top_n
        )
    else:
        print("  [5/6] Skipping reranker (use_reranker=False)...")

    # 6. Build graph
    print("  [6/6] Compiling graph...")

    def _join(docs):
        return "\n\n".join(d.page_content for d in docs)

    def _rewrite(state):
        prompt = prompts["rewriter"].format(question=state["question"])
        return {"rewritten_question": _llm_text(cfg_llm.invoke(prompt)).strip()}

    def _retrieve(state):
        retriever = cfg_vectorstore.as_retriever(
            search_kwargs={"k": config.retriever_k}
        )
        query = state.get("rewritten_question", state["question"])
        docs = retriever.invoke(query)
        return {
            "retrieved_docs": docs,
            "retrieved_context": _join(docs),
            "source_documents": [d.metadata for d in docs],
        }

    def _rerank(state):
        docs = state.get("retrieved_docs", [])
        if not docs or cfg_compressor is None:
            return {}
        reranked = list(
            cfg_compressor.compress_documents(docs, query=state["question"])
        )
        return {
            "retrieved_docs": reranked,
            "retrieved_context": _join(reranked),
            "source_documents": [d.metadata for d in reranked],
        }

    def _reader(state):
        prompt = prompts["reader"].format(
            question=state["question"], context=state.get("retrieved_context", "")
        )
        return {"extracted_facts": _llm_text(cfg_llm.invoke(prompt)).strip()}

    def _research(state):
        prompt = f"{prompts['system']}\n\n" + prompts["research"].format(
            question=state["question"], context=state.get("extracted_facts", "")
        )
        return {"research_answer": _llm_text(cfg_llm.invoke(prompt)).strip()}

    def _marketing(state):
        prompt = f"{prompts['system']}\n\n" + prompts["marketing"].format(
            question=state["question"], context=state.get("extracted_facts", "")
        )
        return {"marketing_answer": _llm_text(cfg_llm.invoke(prompt)).strip()}

    b = StateGraph(DualPersonaState)
    b.add_node("retrieve_context", _retrieve)
    b.add_node("reader", _reader)
    b.add_node("research", _research)
    b.add_node("marketing", _marketing)
    b.add_node("merge_outputs", merge_outputs_node)

    if config.use_rewriter:
        b.add_node("rewrite", _rewrite)
        b.add_edge(START, "rewrite")
        b.add_edge("rewrite", "retrieve_context")
    else:
        b.add_edge(START, "retrieve_context")

    if config.use_reranker:
        b.add_node("rerank_context", _rerank)
        b.add_edge("retrieve_context", "rerank_context")
        b.add_edge("rerank_context", "reader")
    else:
        b.add_edge("retrieve_context", "reader")

    b.add_edge("reader", "research")
    b.add_edge("reader", "marketing")
    b.add_edge("research", "merge_outputs")
    b.add_edge("marketing", "merge_outputs")
    b.add_edge("merge_outputs", END)

    compiled = b.compile()

    components = {
        "config": config,
        "prompts": prompts,
        "vectorstore": cfg_vectorstore,
        "llm": cfg_llm,
        "model": cfg_model,
        "compressor": cfg_compressor,
        "graph": compiled,
    }
    print(f'Pipeline "{config.name}" ready.\n')
    return compiled, components
```

### Main EVAL

```python
def run_full_evaluation(
    configs: list[RAGConfig],
    dataset: dict = None,
    scorer=None,
    verbose: bool = True,
) -> dict:
    """Build and evaluate one or more pipeline configs."""
    if dataset is None:
        dataset = validation_questions_answers
    if scorer is None:
        scorer = bert_scorer

    all_results = {}

    for config in configs:
        if verbose:
            print("\n" + "#" * 70)
            print(f"# EVALUATING: {config.summary()}")
            print("#" * 70)

        graph, components = build_rag_pipeline(config)

        t0 = time.time()
        metrics, raw_rows = evaluate_rag(dataset, scorer=scorer, graph=graph)
        elapsed = time.time() - t0
        n = max(len(dataset), 1)

        if verbose:
            print(
                f"\nTotal eval time: {elapsed:.1f}s ({n} questions, {elapsed / n:.1f}s/question)"
            )

        all_results[config.name] = {
            "metrics": metrics,
            "rows": raw_rows,
            "time": elapsed,
            "config": config,
        }

        # Free GPU memory between configs
        components.pop("model", None)
        components.pop("llm", None)
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    if len(configs) > 1 and verbose:
        print("\n\n" + "=" * 80)
        print("COMPARISON SUMMARY")
        print("=" * 80)
        header = f"{'Config':<24} {'Persona':<12} {'BERT-F1':>9} {'ROUGE-L':>9} {'Faith':>9} {'Len-R':>7} {'Time':>8}"
        print(header)
        print("-" * len(header))
        for config_name, res in all_results.items():
            t = f"{res['time']:.1f}s"
            for persona in ["Research", "Marketing"]:
                m = res["metrics"][persona]
                bf1 = f"{m['bertscore'][2]:.4f}"
                rl = f"{m['rouge'][2]:.4f}"
                faith = (
                    f"{m['faithfulness_f1']:.4f}"
                    if m["faithfulness_f1"] is not None
                    else "N/A"
                )
                lr = f"{m['length_ratio']:.2f}"
                print(
                    f"{config_name:<24} {persona:<12} {bf1:>9} {rl:>9} {faith:>9} {lr:>7} {t:>8}"
                )
                t = ""

    return all_results


def results_to_dataframe(all_results: dict, include_text: bool = False):
    """
    Convert evaluation results into a tidy DataFrame.

    Parameters
    ----------
    include_text : bool
        If True, add columns for each question's generated research/marketing
        answers alongside their per-question BERT-F1 and ROUGE-L scores.
    """
    # ── aggregate rows ────────────────────────────────────────────────────
    agg_rows = []
    for config_name, res in all_results.items():
        elapsed = res["time"]
        for persona, m in res["metrics"].items():
            agg_rows.append(
                {
                    "config": config_name,
                    "persona": persona,
                    "bertscore_f1": round(m["bertscore"][2], 4),
                    "rouge_l": round(m["rouge"][2], 4),
                    "faithfulness_f1": round(m["faithfulness_f1"], 4)
                    if m["faithfulness_f1"] is not None
                    else None,
                    "length_ratio": round(m["length_ratio"], 2),
                    "eval_time_sec": round(elapsed, 1),
                }
            )

    df_agg = pd.DataFrame(agg_rows).sort_values(
        by=["persona", "bertscore_f1", "rouge_l"], ascending=[True, False, False]
    )

    if not include_text:
        return df_agg

    # ── per-question text rows ────────────────────────────────────────────
    text_rows = []
    for config_name, res in all_results.items():
        for row in res.get("rows", []):
            text_rows.append(
                {
                    "config": config_name,
                    "q_id": row["id"],
                    "question": row["question"],
                    "gold_research": row["gold_research"],
                    "gen_research": row["gen_research"],
                    "bert_f1_research": round(row["bert_f1_research"], 4),
                    "rouge_l_research": round(row["rouge_l_research"], 4),
                    "gold_marketing": row["gold_marketing"],
                    "gen_marketing": row["gen_marketing"],
                    "bert_f1_marketing": round(row["bert_f1_marketing"], 4),
                    "rouge_l_marketing": round(row["rouge_l_marketing"], 4),
                }
            )

    df_text = pd.DataFrame(text_rows)

    pd.set_option("display.max_colwidth", 300)
    return df_agg, df_text
```

```python
def run_test_question_report(
    test_dataset: dict,
    config,
    metric_ids: tuple = (),
    scorer=None,
):
    """
    Run each question in `test_dataset` through a fresh pipeline built from
    `config` and print the responses needed for the section 5 questions
        a) research + marketing responses
        b) context passed to the LLM
        c) doc_source for each context document
        d) per-persona BERTScore/ROUGE vs gold (only for item_ids in `metric_ids`)

    Pass the gold answers inside `test_dataset[item_id]` as
    `gold_answer_research` / `gold_answer_marketing` so (d) can score them.
    """
    from rouge_score import rouge_scorer as _rs

    _rouge = _rs.RougeScorer(["rouge1", "rouge2", "rougeL"], use_stemmer=True)

    def _bert(pred, ref):
        if scorer is not None:
            p, r, f = scorer.score([pred], [ref])
        else:
            from bert_score import score as bscore

            p, r, f = bscore([pred], [ref], lang="en", verbose=False)
        return float(p.mean()), float(r.mean()), float(f.mean())

    print(f"Building pipeline from config: {config.name}")
    graph, _ = build_rag_pipeline(config)

    results = {}
    for item_id, item in test_dataset.items():
        question = item.get("question", "")
        print("\n" + "=" * 80)
        print(f"TEST QUESTION {item_id}: {question}")
        print("=" * 80)

        state = graph.invoke({"question": question})

        research = state.get("research_answer") or ""
        marketing = state.get("marketing_answer") or ""
        context = state.get("retrieved_context") or ""
        sources = state.get("source_documents") or []

        # (a) responses
        print("\n--- (a.i) RESEARCH RESPONSE ---")
        print(research)
        print("\n--- (a.ii) MARKETING RESPONSE ---")
        print(marketing)

        # (b) context
        print("\n--- (b) CONTEXT PASSED TO LLM ---")
        print(context)

        # (c) sources
        print("\n--- (c) DOCUMENT SOURCES ---")
        doc_sources = []
        for i, d in enumerate(sources):
            md = d if isinstance(d, dict) else (getattr(d, "metadata", {}) or {})
            src = (
                md.get("doc_source")
                or md.get("source")
                or md.get("title")
                or "<unknown>"
            )
            doc_sources.append(src)
            print(f"  [{i}] {src}")

        entry = {
            "question": question,
            "research_answer": research,
            "marketing_answer": marketing,
            "retrieved_context": context,
            "doc_sources": doc_sources,
        }

        # (d) metrics — only for the item_ids the caller asked for
        if item_id in metric_ids:
            gold_res = item.get("gold_answer_research", "")
            gold_mkt = item.get("gold_answer_marketing", "")

            print("\n--- (d.i) RESEARCH METRICS vs GOLD ---")
            br_p, br_r, br_f = _bert(research, gold_res)
            rs_res = _rouge.score(gold_res, research)
            research_metrics = [
                ("bertscore_precision", round(br_p, 4)),
                ("bertscore_recall", round(br_r, 4)),
                ("bertscore_f1", round(br_f, 4)),
                ("rouge1_f", round(rs_res["rouge1"].fmeasure, 4)),
                ("rouge2_f", round(rs_res["rouge2"].fmeasure, 4)),
                ("rougeL_f", round(rs_res["rougeL"].fmeasure, 4)),
            ]
            for name, val in research_metrics:
                print(f"  {name}: {val}")

            print("\n--- (d.ii) MARKETING METRICS vs GOLD ---")
            bm_p, bm_r, bm_f = _bert(marketing, gold_mkt)
            rs_mkt = _rouge.score(gold_mkt, marketing)
            marketing_metrics = [
                ("bertscore_precision", round(bm_p, 4)),
                ("bertscore_recall", round(bm_r, 4)),
                ("bertscore_f1", round(bm_f, 4)),
                ("rouge1_f", round(rs_mkt["rouge1"].fmeasure, 4)),
                ("rouge2_f", round(rs_mkt["rouge2"].fmeasure, 4)),
                ("rougeL_f", round(rs_mkt["rougeL"].fmeasure, 4)),
            ]
            for name, val in marketing_metrics:
                print(f"  {name}: {val}")

            entry["research_metrics"] = research_metrics
            entry["marketing_metrics"] = marketing_metrics

        results[item_id] = entry

    return results
```
